# Smith Research Analysis

Synthetic regression analysis (ML / research only). Same analysis as the Deepnote notebook. No real money, no mining.

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

rng = np.random.default_rng(42)
X = rng.normal(size=(200, 4))
y = 60 * X[:, 0] + 3.5 * X[:, 1] + 33 * X[:, 2] + 84 * X[:, 3] + rng.normal(scale=8, size=200)
df = pd.DataFrame(X, columns=["f1", "f2", "f3", "f4"])
df["target"] = y
model = LinearRegression().fit(X, y)
pred = model.predict(X)
print("Coefficients", model.coef_)
print("MSE", round(mean_squared_error(y, pred), 2))
print("R2", round(r2_score(y, pred), 4))
print("ML research analysis complete.")
```
